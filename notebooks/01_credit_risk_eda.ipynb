```python
from pathlib import Path
import pandas as pd

# Look for data folder relative to notebook, repo root, or original home folder
possible_folders = [
    Path("../data"),
    Path("data"),
    Path.home() / "Documents" / "CreditRisk-AI" / "data"
]
data_folder = next((p for p in possible_folders if p.exists()), None)
if not data_folder:
    raise FileNotFoundError("Dataset folder not found. Please ensure the data/ folder exists.")

file_path = next(data_folder.rglob("*.xls"))

print("Dataset found at:")
print(file_path)

df = pd.read_excel(
    file_path,
    header=1,
    engine="xlrd"
)

print("
Dataset Shape:", df.shape)
df.head()
```
